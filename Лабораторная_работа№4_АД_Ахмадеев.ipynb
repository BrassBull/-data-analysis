**<p>1. Задача 2.25 </p>**
<p>12 очков можно набрать только в 1 из 36 исходов, вероятность ни разу не выбить 12 очков за n бросков: $(\frac{35}{36})^n$</p>
<p>Тогда искомая вероятность: $1 - (\frac{35}{36})^n$</p>

```python
import random
import numpy as np
random.seed(a=42, version=2)


# Зададим число эксперементов и несколько разных n
exps = 100000
n = [10, 100, 1000]
# По аналитической формуле P(n1) = 0.246, P(n2) = 0.941, P(n3) = 1 (с округлением)

# Пространство элементарных событий
A = list(range(1,7))
ans = []

# Бросаем два кубика
for j in n:
  k_success=0
  for exp in range(exps):
    for k in range(j):
      dice = random.choices(A, k=2)
      if dice[0] + dice[1] == 12:
          k_success += 1
          break
  ans.append(k_success/exps)

# Методом Монте-Карло получили P(n1), P(n2), P(n3):
print(np.round(ans,4))
```

**<p>3. Задача 3.12 </p>**
<p>а) Уравнение имеет действительные корни когда дискриминант не отрицательный, вероятность этого считаем по интегралу: $P_1 = \int_{0}^{1} \frac{x^2}{4}dx = \frac{1}{12} \approx 0.8333$</p>
<p>b) Так как $\xi, \eta > 0$, то $P_2 = 0$</p>
<p>c) Так как $\xi, \eta > 0$, то $P_3 = 0$</p>

```python
import random
random.seed(a=42, version=2)
# a)

exps = 300000
k_success=0

for exp in range(exps):
    a,b = random.random(), random.random() # берем точку из uniform([0,1]*[0,1])
    roots = np.roots([1,a,b])
    if not np.any(np.imag(roots) > 0): # проверяем корни
        k_success += 1

p = k_success/exps


# Методом Монте-Карло получили:
print(np.round(p,4))
```

```python
# b)

exps = 300000
k_success=0

for exp in range(exps):
    a,b = random.random(), random.random()
    roots = np.roots([1,a,b])
    if not np.any(np.imag(roots) > 0):
        if np.all(roots > 0):
          k_success += 1

p = k_success/exps


# Методом Монте-Карло получили:
print(np.round(p,4))
```

```python
# c)

exps = 300000
k_success=0

for exp in range(exps):
    a,b = random.random(), random.random()
    roots = np.roots([1,a,b])
    if not np.any(np.imag(roots) > 0) and len(roots) > 1:
        if roots[0] * roots[1] < 0:
          k_success += 1

p = k_success/exps


# Методом Монте-Карло получили:
print(np.round(p,4))
```
