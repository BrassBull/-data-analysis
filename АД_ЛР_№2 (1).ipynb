Задача 1: проверить гипотезу при уровне значимости 0.05

H0: win rate на глине и без нее одинаковый

H1: различается

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

# Данные
# Глина (A)
wins_A = 1660
total_A = 2523

# Не глина (B)
wins_B = 3658
total_B = 6373

successes = np.array([wins_A, wins_B])
trials = np.array([total_A, total_B])
```

```python
# z-test
stat, p_value = proportions_ztest(successes, trials)

print(f"Z-статистика: {stat:.4f}")
print(f"P-value: {p_value:.6f}")
```

```python
# Конверсии
rate_A = wins_A / total_A
rate_B = wins_B / total_B

print(f"Win rate (глина): {rate_A:.4f}")
print(f"Win rate (не глина): {rate_B:.4f}")
print(f"Разница: {rate_A - rate_B:.4f}")
```

Вывод: различие винрейта на глине статистически значимо.

Задача 2: проверить гипотезу

H0: win rate при подаче и приеме одинаковый

H1: различается

```python
# Данные
# Подача (A)
wins_A = 3837
total_A = 4454

# Приём (B)
wins_B = 1481
total_B = 4442

successes = np.array([wins_A, wins_B])
trials = np.array([total_A, total_B])
```

```python
# z-test
stat, p_value = proportions_ztest(successes, trials)

print(f"Z-статистика: {stat:.4f}")
print(f"P-value: {p_value:.10f}")
```

```python
# Конверсии
rate_serve = wins_A / total_A
rate_return = wins_B / total_B

print(f"Win rate на подаче: {rate_serve:.4f}")
print(f"Win rate на приёме: {rate_return:.4f}")
print(f"Разница: {rate_serve - rate_return:.4f}")
```

Вывод: различие винрейта на подаче и приеме статистически значимо.
